# Project 9 — Rubric-based auto-grader
**Track A (local Ollama).** Build an LLM-as-judge and calibrate it against human labels.
**Data:** `data/graded_answers.jsonl` — 15 answers with a human score (0-3).
**Evaluated on:** agreement between the judge and the human scores.

```python
import sys, json; sys.path.append("../..")   # import utils/ and eval/ from repo root
from utils import ask

GRADED = [json.loads(l) for l in open("data/graded_answers.jsonl", encoding="utf-8")]
print(len(GRADED), "graded answers — example:", GRADED[0])
```

## Starter judge

```python
RUBRIC = "Score 0-3: 0 wrong, 1 vague, 2 mostly right, 3 correct and clear."

def grade(question, answer):
    prompt = (f"Question: {question}\nRubric: {RUBRIC}\n"
              f'Answer: "{answer}"\n\n'
              'Return ONLY JSON {"score": int, "reason": str}.')
    raw = ask(prompt)
    try:
        return json.loads(raw)["score"]
    except Exception:
        return None

r = GRADED[0]
print(r["question"], "| human:", r["human_score"], "| judge:", grade(r["question"], r["answer"]))
```

## Measure agreement with the human labels

```python
exact_hits = within1 = n = 0
for r in GRADED:
    js = grade(r["question"], r["answer"])
    if js is None:
        continue
    n += 1
    exact_hits += (js == r["human_score"])
    within1 += (abs(js - r["human_score"]) <= 1)
print(f"exact agreement: {exact_hits}/{n} | within +/-1: {within1}/{n}")
```
