# Project 1 — Support-ticket triage bot
**Track A (local Ollama).** Classify a ticket, extract fields, draft a reply.
**Data:** `data/tickets.jsonl` — 24 tickets labelled with `label`, `product`, `severity`.
**Evaluated on:** classification accuracy (+ extend to field extraction). **Injection test included.**

```python
import sys, json; sys.path.append("../..")   # import utils/ and eval/ from repo root
from utils import ask
from eval import run_eval, print_report, exact

TICKETS = [json.loads(l) for l in open("data/tickets.jsonl", encoding="utf-8")]
print(len(TICKETS), "tickets loaded — example:", TICKETS[0])
```

## Starter prompt: zero-shot classifier

```python
LABELS = ["billing", "bug", "other"]

def classify(ticket: str) -> str:
    prompt = (
        "Classify the support ticket as billing, bug, or other.\n"
        "Reply with ONE lowercase word and nothing else.\n\n"
        f'Ticket: "{ticket}"'
    )
    out = ask(prompt).strip().lower()
    return next((lab for lab in LABELS if lab in out), out)

print(classify(TICKETS[1]["input"]))
```

## Score it on the dataset

```python
cases = [{"input": t["input"], "expected": t["label"]} for t in TICKETS]
print_report(run_eval(cases, classify, scorer=exact))
```
