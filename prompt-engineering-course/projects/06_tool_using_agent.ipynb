## Tools (the description IS the prompt)

```python
def calculator(expr: str) -> str:
    allowed = set("0123456789+-*/(). ")
    return str(eval(expr)) if set(expr) <= allowed else "error: invalid expression"

def lookup(key: str) -> str:
    return str(FACTS.get(key, "unknown"))

TOOLS = {
    "calculator": {"fn": calculator, "desc": "Evaluate arithmetic. Args: expr (string)."},
    "lookup":     {"fn": lookup,     "desc": "Look up a fact by key. Keys: " + ", ".join(FACTS) + "."},
}
for n, t in TOOLS.items():
    print(n, "-", t["desc"])
```
